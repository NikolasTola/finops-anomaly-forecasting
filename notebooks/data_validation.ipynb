```python
import pandas as pd
```

```python
df = pd.read_csv('../data/raw/ec2_cpu_utilization_24ae8d.csv', parse_dates=['timestamp'])
df = df.set_index('timestamp')
df.head()
```

```python
duplicatas = df.index.duplicated().sum()
print(f'Timestamps duplicados: {duplicatas}')

if duplicatas > 0:
    print(df[df.index.duplicated(keep=False)])
```

```python
esta_ordenado = df.index.is_monotonic_increasing
print(f'ordenado cronologicamente: {esta_ordenado}')
```

```python
diffs = df.index.to_series().diff().dropna()
print('Contagem intervalos:')
print(diffs.value_counts())
```

```python
frequencia_esperada = '5min'
range_completo = pd.date_range(start=df.index.min(), end=df.index.max(), freq=frequencia_esperada)

timestamps_faltantes = range_completo.difference(df.index)
print(f'Timestamps esperados: {len(range_completo)}')
print(f'Timestamps presentes: {len(df)}')
print(f'Gap de imestamps faltantes: {len(timestamps_faltantes)}')

if len(timestamps_faltantes) > 0:
    print(timestamps_faltantes)
```
